# Modifying the global scope

```python
"""DONOT MODIFY IT UNLESS VERY NECESSARY"""
x = 10  # global variable

def modify_global():
    global x
    x = 20  # modifies the global variable
print(x) #before modifying global
modify_global()
print(x)  # prints 20
```

# Prime or not 

```python
def prime_check(number):
    if number <= 1:
        return False
    for i in range(2, int(number**0.5) + 1):   
        if number % i == 0:
            return False
    return True

prime_check(37)
```
