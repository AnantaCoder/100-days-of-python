# Functions


```python
def greet(name: str) -> None:
    """Prints a personalized greeting message"""
    print(f"Hello, {name}!")

# Call the function
greet("John") # Output: Hello, John!
greet("Allu") # Output: Hello, John!
```

# Join Method

```python
number = 18
i=1
num_list = []
while i<=number:
    num_list.append(i)
    i +=1
print(num_list)
#joinin gthe striung too a new line of 
y = "".join([str(j) for j in num_list]) #one line loop 
print(y)
```

# Ascii Converter :

```python
word = input("Enter the letter")
word_list= []
for letter in word:
    ascii = ord(letter)
    print(ascii)
    forward = ascii + 1
    print(forward)
    new_word = chr(forward)
    word_list.append(new_word)
y = "".join([str(new_letter) for new_letter in word_list])
print(y)
    
```

# Map Function

```python
numbers = [1, 2, 3, 4, 5]
squared_numbers = list(map(lambda x: x ** 2, numbers))
print(squared_numbers) 
```
