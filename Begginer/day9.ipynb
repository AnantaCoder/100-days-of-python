# Nested list

```python
# Nested list representing students and their grades
student_grades = [
    ["Alice", [85, 90, 88]],
    ["Bob", [78, 82, 79]],
    ["Charlie", [92, 88, 95]],
    ["David", [70, 75, 72]],
    ["Eva", [88, 91, 85]]
]

# If you want to print a specific student's grades:
# Let's say you want to print Bob's second grade (index 1)
print(student_grades[1][1][1])  # This prints Bob's second grade, which is 82

# Print the nested list
for student in student_grades:
    print(f"Student: {student[0]}, Grades: {student[1]}")
```

# 2d List aka matrix

```python
rows = int(input("Enter the number of rows: "))
columns = int(input("Enter the number of columns: "))
matrix = []  # empty list

for i in range(rows):
    # For each row, the user is prompted to enter elements separated by spaces.
    # The map(int, input().split()) converts the input string into a list of integers.
    row = list(map(int, input(f"Enter the elements for row {i+1}, separated by spaces: ").split()))
    
    # Ensure the row has the correct number of columns
    while len(row) != columns:
        print(f"Please enter exactly {columns} elements.")
        row = list(map(int, input(f"Enter the elements for row {i+1}, separated by spaces: ").split()))
    
    # Add the row to the matrix
    matrix.append(row)

# Print the matrix
print("The matrix is:")
for row in matrix:  # use `row` to iterate over the rows of the matrix
    for element in row:  # use `element` to iterate over elements in each row
        print(element, end=" ")
    print()  # move to the next line after printing each row
```
