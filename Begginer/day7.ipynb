#   converting a list to dictionary

```python
list_of_tuples =  [("name", "John"), ("age", 25), ("city", "New York")]
dictionary = dict(list_of_tuples)
print(dictionary)
```

```python
list_items = ['name', 'John', 'age', 25, 'city', 'New York']

# Use slicing to create pairs (i.e., 2 elements at a time)
dictionary = {list_items[i]: list_items[i+1] for i in range(0, len(list_items), 2)}
print(dictionary)
# Output: {'name': 'John', 'age': 25, 'city': 'New York'}
```

```python
keys = ['name', 'age', 'city']
values = ['John', 25, 'New York']

dictionary = dict(zip(keys, values))
print(dictionary)
# Output: {'name': 'John', 'age': 25, 'city': 'New York'}
```
