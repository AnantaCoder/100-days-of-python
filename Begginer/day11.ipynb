# dictionary 

```python
dic = {
        'name': 'Justin Timberlake',
        'follower_count': 59,
        'description': 'Musician and actor',
        'country': 'United States'
}
print(dic['name'])
for key in dic.keys():
    print(f"The value corresoinding to the key {key} is {dic[key]}")
```

```python
print(dic.items())
```

# Joining two dictionaries

```python
# Define two dictionaries
dict1 = {'a': 1, 'b': 2}
dict2 = {'c': 3, 'd': 4}

# Use the update() method to join dict2 into dict1
dict1.update(dict2)

# Now dict1 contains all key-value pairs from both dictionaries
print(dict1)
dict1.pop('a')
print(dict1)
dict1.popitem() # pops an item
print(dict1)

```
