```python
# List of Indian states
indian_states = [
    "Andhra Pradesh", "Arunachal Pradesh", "Assam", "Bihar", "Chhattisgarh",
    "Goa", "Gujarat", "Haryana"
]

# Print the list
# print(indian_states)
indian_states[2] = "Hawai" #changing the existring list
print(indian_states[2:5])
indian_states.append("Ankorawat") # adding to the list
print(indian_states)
indian_states.pop(3) # Pass 3 directly, not [3]

indian_states.extend(["Raja Mouli" , "Sitaraman"]) #joining two list
print(indian_states)
```

```python
names_list = []  # Initialize an empty list to store names

# Use a loop to input multiple names
while True:
    name = input("Enter the Name (or type 'stop' to finish): ")
    if name.lower() == 'stop':  # Check if the user wants to stop
        break
    names_list.append(name)  # Add the name to the list

# Print the final list of names
print("Names List:", names_list)
```
