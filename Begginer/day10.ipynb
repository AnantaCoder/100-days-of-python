# title

```python
def format_name(name , b_name):
    f_name = name.title()
    s_name = b_name.title()
    return f"{f_name} {s_name}"
    
print(format_name("anioeka", "saSNAVKA"))
```

```python
def format(fname , lname):
    if fname =="" or lname == "":
        return "Invalid input"
    formated_f = fname.title()
    formated_l = lname.title()
    return f"Result: {formated_f} {formated_l}"
print(format_name(input("Whats your first name?") , input("Whats your last name?")))
```

# calender

```python
import calendar

# Input year and month from the user
year = int(input("Enter year: "))
month = int(input("Enter month (1-12): "))

# Display the calendar
cal = calendar.month(year, month)
print(cal)
```

# docstring

```python
def greet(name: str) -> None:
    """Print a personalized greeting message.

    Args:
        name (str): The name of the person to greet.

    Returns:
        nill
    """
    print(f"Hello, {name}!")
greet("Anirban")
```
