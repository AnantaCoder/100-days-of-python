### parameter = name of the data passed in \

```python
def greet(name, location):
    print(f"Hellow {name} 🎉")
    print(f"Hows the weather today in  {location}?")
    
greet("Annamalai", "Kolkata")
    
```

```python
def swap(a, b):
    return b, a

x = 3
y = 5
x, y = swap(x, y)
print(f"X is {x} y is {y}")
```

# Change the ascii value

```python
# Get the ASCII value of a character
char = 'A'
ascii_value = ord(char)

# Modify the ASCII value
new_ascii_value = ascii_value + 1 

# Convert the new ASCII value back to a character
new_char = chr(new_ascii_value)

print(new_char)
```
