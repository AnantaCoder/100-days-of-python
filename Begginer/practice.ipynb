```python
from typing import List

class Solution:
    def removeElement(self,nums:List[int], target: int)->List[int]:
        k = 0
        for i in range(len(nums)):
            if nums[i] != target:
                nums[k] = nums[i]
                k +=1 
        return k 
    
sol = Solution()
nums = [3,2,2,3]
last = nums[len(nums)-1]
target = 3
k = sol.removeElement(nums,target)
print(k)
print(nums[:k])
print(last)
```

```python
from typing import List
class Solution:
    def plusOne(self, digits: List[int]) -> List[int]:
        for i in range(len(digits) - 1, -1, -1):
            
            if digits[i] == 9:
                digits[i] = 0
                if i > 0 and digits[i - 1] != 9:
                    digits[i - 1] += 1
                    return digits
            else:
                digits[i] += 1
                return digits
        
        
        return [1] + digits

sol = Solution()
digits = [1, 9, 9, 9]
new = sol.plusOne(digits)
print(new)
```
