# Handeling the json files 


```python
import requests
import html
parameters={
    "amount":20,
    "type":"multiple"
}
# Send a GET request to the Open Trivia Database API
response = requests.get('https://opentdb.com/api.php?', params=parameters)

# Parse the response as JSON
data = response.json()

# Access the first question's correct and incorrect answers
question_data = data['results'][0]
correct_answer = question_data['correct_answer']
q = html.unescape(question_data['question'])
incorrect_answers = question_data['incorrect_answers']

# Print the correct and incorrect answers
print(q)
print("Correct Answer:", correct_answer)
print("Incorrect Answers:", incorrect_answers)
```
