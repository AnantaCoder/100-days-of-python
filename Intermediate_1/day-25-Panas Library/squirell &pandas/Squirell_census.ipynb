# Importing The libraries

```python
import pandas as pd 
dataset = pd.read_csv(r'squirell_database.csv')
x  = dataset.loc[:,"Primary Fur Color"]

gray = 0
black = 0
red = 0
for _ in x:
    if _ == "Gray":
        gray += 1
    elif _ == "Black":
        black += 1
    elif _ == "Cinnamon":
        red +=1
print(black)
print(gray)
print(red)


```

# Creating a new file for the data

```python
Fur_color = {
    'Fur Color' : ['Black','Cinnamon','Gray'],
    'Count' : [black,red,gray]
}
new_file = pd.DataFrame(Fur_color)
new_file.to_csv('Squirell Fur Colr.csv')
```
