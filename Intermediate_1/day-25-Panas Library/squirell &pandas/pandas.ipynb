```python
import pandas as pd
dataset = pd.read_csv("weather_data.csv") 
print(dataset)
x = dataset.iloc[: , 1:-1].values #.vslues to use it to store it in a 2d array
print(x.mean())
```

```python
temp_list = dataset['temp'].to_list()
print(temp_list)
# avgtemp = sum(temp_list)/len(temp_list)
# print(avgtemp)
print(dataset['temp'].max()) #maximum in the datset
print(dataset['temp'].mean()) #mean or average in the dataset
print(dataset.condition) #object 
```

```python
#get data in the rows 
print(dataset[dataset.day == 'Monday'])
print([dataset.day == dataset['temp'].max()])
```
