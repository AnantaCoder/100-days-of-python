# Creating a us state Name Guesser  using Turtle and Pandas Library

```python
import turtle
import pandas

# Load state data
data = pandas.read_csv('50_states.csv')

# Set up the screen and turtle
screen = turtle.Screen()
screen.title("U.S. States Game")
image = 'blank_states_img.gif'
screen.addshape(image)
turtle.shape(image)

# Create a turtle for writing text
t = turtle.Turtle()
t.hideturtle()
t.penup()

# Initialize guessed states and states list
guessed_states = []
states_list = data.state.to_list()

while len(guessed_states) < 50:
    # Get user input
    answer = screen.textinput(title=f"{len(guessed_states)}/50 States Guessed", prompt="Name a State")
    
    if answer is None:
        # Handle cancel case
        print("User cancelled the input.")
        continue
    
    answer = answer.title()
    
    if answer == "Exit":
        # Collect missing states and save to CSV
        missing_states = [state for state in states_list if state not in guessed_states]
        new_data = pandas.DataFrame(missing_states, columns=['State'])
        new_data.to_csv("states_to_learn.csv", index=False)
        print("Missing states saved to 'states_to_learn.csv'")
        break  # Exit the loop after saving missing states

    if answer in states_list and answer not in guessed_states:
        guessed_states.append(answer)
        # Find the coordinates of the guessed state
        check_answer = data[data.state == answer]
        x_axis, y_axis = int(check_answer.x.values[0]), int(check_answer.y.values[0])
        t.goto(x_axis, y_axis)
        t.write(answer)

# Close the screen on click
screen.mainloop()
```
