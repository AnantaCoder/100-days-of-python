```python
import random
names = ['anirban','dave', 'akash ' , 'amit nigam']
student_score = {student:random.randint(50,100)  for student in names}
print(student_score)
```

```python

passed = {student:score for (student, score) in student_score.items() if score>=75}
print(passed)
```

```python
sentence = "What is the Airspeed Velocity of an Unladen Swallow?"
result_list = {word: len(word) for word in sentence.split()}
print(result_list)
```

```python
weather_c = {"Monday": 12, "Tuesday": 14, "Wednesday": 15, "Thursday": 14, "Friday": 21, "Saturday": 22, "Sunday": 24}

weather_f =  {day: (temp_c * 9/5) + 32 for day, temp_c in weather_c.items()}

print(weather_f)
```
