```python
students_marks = {
    "Alice": 85,
    "Bob": 92,
    "Charlie": 78,
    "Diana": 95,
    "Evan": 88
}

for student, mark in students_marks.items():
    print(f"{student}: {mark}")
```

```python
import pandas as pd

# Parse the log
log = """
Alice : 85
Bob : 92
Charlie : 78
Diana : 95
Evan : 88
"""

students_marks = {}
for line in log.strip().split('\n'):
    name, mark = line.split(' : ')
    students_marks[name] = int(mark)

# Create a Pandas DataFrame
students_marks_df = pd.DataFrame(list(students_marks.items()), columns=['Student', 'Mark'])

print(students_marks_df)
```

```python
import pandas as pd

# Create a dataframe
data = {
    'name': ['Mike', 'Doe', 'James'],
    'age': [18, 19, 29]
}
df = pd.DataFrame(data)

# Loop through the rows using iterrows()
for index, row in df.iterrows():
    if row['name'] == 'Mike':
        print(f"Mike's age is: {row['age']}")

# Optional: Print the entire DataFrame
print("\nEntire DataFrame:")
print(df)
```
