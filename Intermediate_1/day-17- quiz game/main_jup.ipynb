# Classes and Objects 

```python
class User:
    def __init__(self, name, age):
        self.name = name 
        self.age = age
        
user_1 = User("anirban " , "120")
print(user_1.age)
```

adding methods to the class , function attached to a class = method 

```python
class Car:
    def __init__(self, name, price):
        self.name = name 
        self.price = price
        self.followers = 0
        self.following = 0
        
    def follow(self,name):
        self.following += 1
        name.followers +=1
        
user_1 = Car("Supra" , "120m")
user_2 = Car("GTR " , "120m")
print(user_1.price)

user_1.follow(user_2)
print(user_1.followers)
print(user_1.following)
print(user_2.followers)
print(user_2.following)
```
