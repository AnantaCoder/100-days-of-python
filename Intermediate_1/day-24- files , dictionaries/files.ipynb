> Reading the file 

```python
file = open('data.txt') 
contents = file.read()
print(contents)
file.close() # to close the file manually 
```

```python
with open('data.txt')  as fil :
    contents =list( fil.read())
    print(contents)
# closes automatically 
```

```python
with open(r'data.txt', mode='w') as file:
    file.write('This is the new text ')
    # this code deletes everything and rewrite everything 
```
