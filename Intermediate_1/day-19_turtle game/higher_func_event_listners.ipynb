
# Higher order functions and a event listner 

```python
def add(x,y):
    return x+y
def sub(x,y):
    return x-y
def mul(x,y):
    return x*y
def div(x,y):
    return x/y
def calculator(x,y , func):
    return func(x,y)

result = calculator(2,3,div)
print(result)
```
