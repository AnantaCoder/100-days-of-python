# Keyword arguments 

```python
def func(a =1 , b=2 ,c=3):
    return a+b+c
print(func(2)) # predefined argument is given 
```

```python
def test(*args):
    print(args)
 
test(1,2,3,5)
```

```python
def foo(a, b=4, c=6):
    print(a, b, c)
 
foo(20, c=6)
```

# unlimited arguments - tuple

```python
def add(*args):
    print(args[1])
    total = 0
    for n in args:
        total += n
    return total
add(1,2,3,4)
```

# many keyword argument - dictionary

```python
def calculate(n,**kwargs):
    print(kwargs)
    # for key , value in kwargs.items():
    #     print(key)
    #     print(value)
    n+=kwargs['add']
    n*=kwargs['multiply']
    print(n)
    
calculate(2,add= 2 , multiply=5)
```

```python
class Car:
    def __init__(self, **kw):
        self.make = kw.get('make')
        self.model = kw.get('model', 'Unknown')  # default value for model
        self.color = kw.get('color')

mycar = Car(make='Nissan',color = 'red')
print(mycar.model)  
print(mycar.color)  
```

```python

def bar(spam, eggs, toast='yes please!', ham=0):
    print(spam, eggs, toast, ham)
 
bar(toast='nah', spam=4, eggs=2)
```

```python
def all_aboard(a, *args, **kw): 
    print(a, args, kw)
 
all_aboard(4, 7, 3, 0, x=10, y=64)
```
