```python
def outer_function():
    outer_var = "Hello from outer function"

    def inner_function():
        print(outer_var)  # Inner function can access outer_var

    inner_function()  # Call the inner function inside the outer one

outer_function()  # Call the outer function
```

```python
import time

def delay_decorative_func(function):
    def wrapper(*args, **kwargs):
        function()
        time.sleep(3)
        function()
        return function(*args, **kwargs)
    return wrapper

@delay_decorative_func
def say_hello():
    print("Hello, world!")

say_hello()
```
