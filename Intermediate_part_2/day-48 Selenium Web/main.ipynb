# using selenium to create a web scraper 

```python
#amazon price tracker 
from selenium import webdriver
from selenium.webdriver.common.by import By
#keep browser open 
chrome_options = webdriver.ChromeOptions()
chrome_options.add_experimental_option("detach", True)
driver = webdriver.Chrome(options=chrome_options)
driver.get('https://www.amazon.in/Boult-Wireless-Earbuds-Playtime-Breathing/dp/B0CVLB99YR/ref=lp_1388921031_1_6_sspa?keywords=Headphones&pf_rd_p=9e034799-55e2-4ab2-b0d0-eb42f95b2d05&pf_rd_r=32XDXRW3ZSGAPHDTDF5H&sp_csd=d2lkZ2V0TmFtZT1zcF9hcGJfZGVza3RvcF9icm93c2VfaW5saW5lX210Zg&th=1')
price = driver.find_element(By.CLASS_NAME , value='a-price-whole').text
symbol = driver.find_element(By.CLASS_NAME , value='a-price-symbol').text

print(f"Price is {symbol}{price}")

#find the element by names , id , classname etc 
search_bar = driver.find_element(By.NAME , value='field-keywords')
print(search_bar.tag_name)
#css sellector use when u dont know the class name etc 
link = driver.find_element(By.CSS_SELECTOR ,value='.nav-progressive-content a')
print(link.text)
# if u cant find it u can use tthe xpath 
xpath = driver.find_element(By.XPATH , value='//*[@id="nav-xshop"]/a[4]')
print(xpath.text)
# we can use find elements which locate every element in the web page 
#driver.close()
driver.quit()
```

# scrap python.org data using selenium

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
#keep open the browser 
chrome_options = webdriver.ChromeOptions()
chrome_options.add_experimental_option('detach' , True)
driver = webdriver.Chrome(options=chrome_options)
driver.get('https://www.python.org')


date = driver.find_elements(By.XPATH , value='//*[@id="content"]/div/section/div[3]/div[2]/div/ul/li/time')
event = driver.find_elements(By.XPATH , value='//*[@id="content"]/div/section/div[3]/div[2]/div/ul/li/a')

# event_list = []
# date_list  = []
# for i in date:
#     date_list.append(i.text)
    
# date_list = [i.text for i in date]

# print(date_list)
# for j in event:
#     event_list.append(j.text)

# event_list = [j.text for j in event]
# print(event_list)
# #merging two lists into a dictionary 
# result_dict = dict(zip(date_list , event_list))
# print(result_dict)
events = {}
for n in range(len(event)):
    events[n] = {
        'date': date[n].text,
        'event': event[n].text
    }
print(events)

driver.quit()
```
